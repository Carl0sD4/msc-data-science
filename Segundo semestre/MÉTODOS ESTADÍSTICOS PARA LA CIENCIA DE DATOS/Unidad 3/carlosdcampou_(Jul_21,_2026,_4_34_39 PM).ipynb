```python
%%R

# Matriz de pagos
pagos <- matrix(c(
  12, 18, 24, 28,
   5, 22, 35, 48,
  -8, 20, 50, 70,
 -25, 10, 55, 95
), nrow = 4, byrow = TRUE)
```

```python
%%R

alternativas <- c("Mercado local", "Expansión regional",
                   "Plataforma nacional en línea", "Expansión internacional")
estados <- c("Contracción", "Crecimiento débil", "Crecimiento moderado", "Crecimiento fuerte")
rownames(pagos) <- alternativas
colnames(pagos) <- estados
```

```python
%%R
# Maximin
minimos <- apply(pagos, 1, min)
maximin <- max(minimos)
cat("Mínimos por alternativa:\n"); print(minimos)
cat("Maximin =", maximin, "->", names(which.max(minimos)), "\n\n")
```

```python
%%R
# Maximax
maximos <- apply(pagos, 1, max)
maximax <- max(maximos)
cat("Máximos por alternativa:\n"); print(maximos)
cat("Maximax =", maximax, "->", names(which.max(maximos)), "\n\n")
```

```python
%%R
# Matriz de arrepentimientos
mejores_columna <- apply(pagos, 2, max)
arrepentimientos <- sweep(-pagos, 2, mejores_columna, "+")
cat("Matriz de arrepentimientos:\n"); print(arrepentimientos)
```

```python
%%R
# Minimax de arrepentimiento
max_arrep <- apply(arrepentimientos, 1, max)
minimax_regret <- min(max_arrep)
cat("\nArrepentimiento máximo por alternativa:\n"); print(max_arrep)
cat("Minimax de arrepentimiento =", minimax_regret, "->", names(which.min(max_arrep)), "\n")
```

```python
%%R

# Datos de la demanda
demanda <- c("Baja", "Media", "Alta", "Excepcional")

#datos de la probabilidad
prob <- c(0.2, 0.35, 0.3, 0.15)
```

```python
%%R
#datos de los pagos
pagos <- matrix(c(
  36, 42, 48, 50,
  20, 50, 70, 82,
 -15, 30, 85, 120
), nrow = 3, byrow = TRUE)
```

```python
%%R
#crear el vector de los planes
planes <- c("Plan pequeño", "Plan mediano", "Plan grande")
rownames(pagos) <- planes
colnames(pagos) <- demanda
```

```python
%%R

#probabilidad sumada solo para verificar que todo haya quedado bien
cat("suma de probabilidades:", sum(prob), ("\n"))

#verificar que las probabilidades mantengan el rango establecido
cat("¿Todas entre 0 y 1?:", all(prob >= 0 & prob <= 1), "\n\n")
```

```python
%%R

# Calcular VEM de cada plan (mostrando cada término ponderado)
for (i in 1:nrow(pagos)) {
  terminos <- pagos[i, ] * prob
  cat(planes[i], "- términos ponderados:", round(terminos, 3), "\n")
  cat(planes[i], "- VEM =", sum(terminos), "\n\n")
}
```

```python
%%R

vem <- pagos %*% prob
cat("VEM de cada plan:\n")
print(vem)
cat("\nMejor alternativa:", planes[which.max(vem)], "con VEM =", max(vem), "\n")
```

# problema 5

```python
%%R

#vector demanda
prob <- c(0.15, 0.35, 0.30, 0.20)
```

```python
%%R

#vector de los pagos

pagos <- matrix(c(
  24, 24, 24, 24,
  12, 36, 36, 36,
  -4, 24, 48, 48,
 -20, 12, 36, 60
), nrow = 4, byrow = TRUE)
```

```python
%%R

#vector de los pedidods
pedidos <- c("Pedir 100", "Pedir 150", "Pedir 200", "Pedir 250")
```

```python
%%R

#calculo de VEN
vem <- pagos %*% prob
rownames(vem) <- pedidos
cat("VEM de cada pedido:\n")
print(vem)
cat("\nMejor decisión sin información:", pedidos[which.max(vem)], "con VEM =", max(vem), "\n")
```

```python
%%R

mejor_por_demanda <- apply(pagos, 2, max)
cat("Mejor pago posible por nivel de demanda:\n")
print(mejor_por_demanda)
```

```python
%%R

valor_con_info_perfecta <- sum(mejor_por_demanda * prob)
cat("\nValor esperado con información perfecta =", valor_con_info_perfecta, "\n")
```

```python
%%R

mejor_vem_sin_info <- max(vem)
veip <- valor_con_info_perfecta - mejor_vem_sin_info
cat("VEIP =", veip, "\n")
```

```python
%%R

costo_informe <- 7.5
cat("¿Conviene comprar el informe? (VEIP > costo):", veip > costo_informe, "\n")

valor_neto_si_compra <- valor_con_info_perfecta - costo_informe
cat("Valor esperado neto si se compra el informe =", valor_neto_si_compra, "\n")
```
