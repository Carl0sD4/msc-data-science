2. (1.0) Hemos asignado a 𝑥 la lista [1, 2, 3] y ahora queremos asignar a 𝑦 una
copia. Podríamos hacer 𝑦 = 𝑥[:], pero parece que 𝑦 = 𝑥 + [] también
funciona. ¿Es así?¿Por qué?

````python
x = [1, 2, 3]
y = x[:]
print(id(x))
print(id(y))
````

````python
z = x + []
print(id(x))
print(id(z))
````

````python
x = [1, 2, 3]
y = x[:]
y.append(99)
z.append(77)
print(x)
print(y)
print(z)
````

Sí, y = x + [] funciona igual que y = x[:], sí crea una copia.

Lo comprobé con id(): id(x), id(y) y id(z) dieron números diferentes, o sea que son objetos distintos en memoria, no el mismo. Además, cuando le hice append() a y y a z, x no cambió, siguió siendo [1, 2, 3]. Eso confirma que son copias independientes y no solo otra referencia a x.

La razón es que tanto x[:] como x + [] crean una lista nueva en vez de devolver la misma lista original, por eso las dos funcionan como copia aunque sean operaciones distintas.
