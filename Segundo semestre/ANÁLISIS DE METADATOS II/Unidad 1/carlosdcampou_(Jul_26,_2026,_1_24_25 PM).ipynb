3. (1.25) ¿Qué valor tiene $z$ tras evaluar estas sentencias?

```
>>> z = 2
>>> z += 2
>>> z += 2 - 2
>>> z *= 2
>>> z *= 1 + 1
>>> z //= 2
>>> z /= 3 - 1
>>> z -= 2 + 1
>>> z -= 2
>>> z **= 3
>>> z
```

````python
#tercer ejercicio

z = 2
print("z =", z)

z += 2
print("z += 2  ->", z)

z += 2 - 2
print("z += 2 - 2  ->", z)

z *= 2
print("z *= 2  ->", z)

z *= 1 + 1
print("z *= 1 + 1  ->", z)

z //= 2
print("z //= 2  ->", z)

z /= 3 - 1
print("z /= 3 - 1  ->", z)

z -= 2 + 1
print("z -= 2 + 1  ->", z)

z -= 2
print("z -= 2  ->", z)

z **= 3
print("z **= 3  ->", z)

print("\nel valor final de z es", z)
````
