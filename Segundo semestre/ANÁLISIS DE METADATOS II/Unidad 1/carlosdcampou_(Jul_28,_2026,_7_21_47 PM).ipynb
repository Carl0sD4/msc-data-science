# 1. (1.25) ¿Qué resulta de evaluar estas expresiones?bold text
```python
>>> str(2.1) + str(1.2)
>>> int(str(2) + str(3))
>>> str(int(12.3)) + '0'
>>> int('2' + '3')
>>> str(int(2.1) + float(3))
```

````python
print(str(2.1) + str(1.2)) #une dos cadenas de texto
print(int(str(2) + str(3))) #otras dos cadenas de texto
print(str(int(12.3)) + '0') #un numero entero mas una cadena de texto, que es cero, no como numero
print(int('2' + '3')) #puede que los declare con entero pero estan dentro de '' y esto hace que sean cadena
print(str(int(2.1) + float(3))) #entero mas (que estaba en decimal) un decimal 3.0, por eso da 5.0
````
