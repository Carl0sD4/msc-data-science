# 3. (3.0) Encuentre los valores numéricos de las siguientes expresiones:
1. $1 + 2(3 + 4)$

2. $4^3 + 3^{2+1}$

3. $\sqrt{(4 + 3)(2 + 1)}$

4. $\left(\dfrac{1+2}{3+4}\right)^2$

```python
#primer ejercicio

#importar la libreria
import math


resultado_primer_ejercicio = 1 + 2 * (3 + 4)
print("el resultado del primer ejecicio es", resultado_primer_ejercicio)
```

```python
#segundo ejercicio

resultado_segundo_ejercicio = 4**3 + 3**(2+1)
print("el resultado del segundo ejercicio es", resultado_segundo_ejercicio)
```

```python
#3er ejercicio

resultado_tercer_ejercicio = math.sqrt((4 + 3) * (2 + 1))
print("resultado del tercer ejercicio es", resultado_tercer_ejercicio)
```

```python
#4to ejercicio

resultado_cuarto_ejercicio = ((1 + 2) / (3 + 4)) ** 2
print("el resultado del cuarto ejercicio es", resultado_cuarto_ejercicio)
```
